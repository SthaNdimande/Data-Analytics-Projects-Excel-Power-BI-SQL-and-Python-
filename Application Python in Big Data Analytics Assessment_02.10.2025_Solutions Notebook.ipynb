### 1. Write a program to display the first 10 prime numbers using the 'while' repetitive structure. [10]

```python
count = 0
num = 0
while count<10:
    is_prime = True
    for i in range(2,int(num**0.5)+1):
        if num%1==0:
            is_prime=False
            if is_prime:
                print(num, end = " ")
    count+=1
```

```python
count = 0
num = 2
while num>1:
    for i in range(2, num):
        if num%i == 0:
            break
    else:
        print(num, end = " ")
        count+=1
    num+=1
    if count==10:
        break
```

### 11. Build a K-Nearest Neighbor model to classify Iphone purchase potential based on Age and Income given that k = 21. Split the data into training and testing sets on a ratio of 75:25. Test the accuracy of the model. [15]

```python
import pandas as pd
ip = pd.read_csv(r"C:\Data\ORT SA\Training\Python\Sets\Iphone Purchase.csv")
ip.head(1)
```

```python
#Select features
x = ip[['Age', 'Salary']]
y = ip[['Purchase Iphone']]
```

```python
#Split the data into training and testing sets
from sklearn.model_selection import train_test_split
x_training, x_testing, y_training, y_testing = train_test_split(x,y, test_size = 0.25)
```

```python
#Normalise the data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_training = scaler.fit_transform(x_training)
x_testing = scaler.transform(x_testing)
```

```python
#Create the classifier
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors = 21)
knn
```

```python
#Train the classifier
knn.fit(x_training, y_training)
```

```python
#Predict Iphone Purchase potential
predicted_ipp = knn.predict(x_testing).round()
predicted_ipp
```

```python
#Test the accuracy of the classifier
from sklearn.metrics import accuracy_score
ac_score = accuracy_score(y_testing, predicted_ipp)
ac_score
```
